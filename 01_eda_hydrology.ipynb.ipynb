```python
import pandas as pd
import numpy as np

monthly_precip = {
    1: 5.2, 2: 3.1, 3: 2.4, 4: 1.2,
    5: 0.8, 6: 0.2, 7: 0.1, 8: 0.3,
    9: 0.5, 10: 1.8, 11: 3.2, 12: 4.9
}

monthly_temp = {
    1: 14.2, 2: 15.8, 3: 19.4, 4: 24.6,
    5: 29.1, 6: 32.5, 7: 33.8, 8: 33.4,
    9: 31.2, 10: 26.5, 11: 20.8, 12: 15.9
}

monthly_evap = {
    1: 45, 2: 52, 3: 78, 4: 105,
    5: 138, 6: 158, 7: 165, 8: 160,
    9: 140, 10: 108, 11: 72, 12: 50
}

stations = [1362100, 1362200, 1362300, 1362400, 1362500]

rows = []
np.random.seed(123)

for station_id in stations:
    for year in range(2000, 2020):
        for month in range(1, 13):
            
            temp_trend = (year - 2000) * 0.03
            noise_precip = np.random.normal(1.0, 0.15)
            noise_temp = np.random.normal(0, 0.5)

            rows.append({
                'basin_id': station_id,
                'date': pd.Timestamp(year=year, month=month, day=1),
                'precip_mm': round(max(monthly_precip[month] * noise_precip, 0), 2),
                'temp_c': round(monthly_temp[month] + temp_trend + noise_temp, 2),
                'evapotranspiration_mm': round(monthly_evap[month] * np.random.normal(1.0, 0.05), 2)
            })

climate_df = pd.DataFrame(rows)
climate_df.to_csv('data/hydrology/nile_climate.csv', index=False)

print("Shape:", climate_df.shape)
print("")
print(climate_df.head(10))
```

```python

station_populations = {
    1362100: 4200000,   # El Ekhsase 
    1362200: 3800000,   # Assiut 
    1362300: 2100000,   # Nag Hammadi
    1362400: 1600000,   # Esna 
    1362500: 980000,    # Gaafra -
}

rows = []
np.random.seed(456)

for station_id, base_pop in station_populations.items():
    for year in range(2000, 2020):
 
        growth = (1.02) ** (year - 2000)
        noise = np.random.normal(1.0, 0.01)

        rows.append({
            'basin_id': station_id,
            'year': year,
            'population_sum': round(base_pop * growth * noise),
        })

population_df = pd.DataFrame(rows)
population_df.to_csv('data/population/nile_population.csv', index=False)

print("Shape:", population_df.shape)
print("")
print(population_df.head(10))
```
